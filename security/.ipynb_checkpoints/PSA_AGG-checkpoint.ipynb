```python
from datetime import datetime
import io
import os
import re
from bs4 import BeautifulSoup
import certifi
from deep_translator import GoogleTranslator
import pandas as pd
import pypdf
import requests
import urllib3

# ---------------------------------------------------------
# 1. Global Setup & SSL Configuration
# ---------------------------------------------------------
os.environ["REQUESTS_CA_BUNDLE"] = certifi.where()
os.environ["SSL_CERT_FILE"] = certifi.where()
urllib3.disable_warnings(urllib3.exceptions.InsecureRequestWarning)

translator = GoogleTranslator(source="en", target="sw")
HEADERS = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36"
}

# ---------------------------------------------------------
# 2. Helper Functions
# ---------------------------------------------------------


def clean_html(raw):
    """Remove HTML tags, strip artifacts, and normalize spaces."""
    if not raw:
        return ""
    text = re.sub(r"<[^>]+>", " ", str(raw))
    text = (
        text.replace("&nbsp;", " ")
        .replace("\u2019", "'")
        .replace("â€“", "–")
        .replace("â€™", "'")
        .replace("â€œ", '"')
        .replace("â€", '"')
    )
    text = re.sub(r"\s+", " ", text).strip()
    if text and not text.endswith("."):
        text += "."
    return text


def to_date(published_at):
    """Format ISO timestamp to YYYY-MM-DD."""
    return (published_at or "").split("T")[0]


def translate_text(text):
    """Translate English sentence into Kiswahili."""
    try:
        if not text or len(text) < 3:
            return ""
        clean_sentence = text.rstrip(".")
        translated = translator.translate(clean_sentence)
        return clean_html(translated)
    except Exception:
        return ""


def get_field(item, field):
    """Extract nested field attributes if present."""
    if "attributes" in item:
        return item["attributes"].get(field, "")
    return item.get(field, "")


def extract_date_from_pdf(doc_url):
    """Extract publication date directly from PDF text using pypdf."""
    try:
        if not doc_url.endswith(".pdf"):
            return ""

        resp = requests.get(doc_url, headers=HEADERS, verify=False, timeout=12)
        if resp.status_code != 200:
            return ""

        pdf_file = io.BytesIO(resp.content)
        reader = pypdf.PdfReader(pdf_file)

        full_pdf_text = ""
        for page in reader.pages:
            full_pdf_text += page.extract_text() or ""

        date_pattern = r"(?:Dated:\s*)?(\b\d{1,2}(?:st|nd|rd|th)?\s+(?:January|February|March|April|May|June|July|August|September|October|November|December)\s+\d{4}\b)"
        match = re.search(date_pattern, full_pdf_text, re.IGNORECASE)

        if match:
            return match.group(1).strip()

        # Fallback: Extract Date from PDF URL structure (e.g. /2024/09/)
        url_match = re.search(r"/(\d{4})/(\d{2})/", doc_url)
        if url_match:
            return f"{url_match.group(1)}-{url_match.group(2)}"

    except Exception as e:
        print(f"   [PDF Date Error] {doc_url}: {e}")

    return ""


# Initialize Master Record Container
all_rows = []

# =========================================================
# SECTION 1: NTSA (Road Safety)
# =========================================================
print("\n[1/4] Fetching data from NTSA...")
try:
    resp_ntsa = requests.get(
        "https://ntsa.go.ke/api/news", headers=HEADERS, verify=False
    )
    items_ntsa = resp_ntsa.json().get("data", [])
    print(f" -> Got {len(items_ntsa)} items from NTSA")

    for item in items_ntsa:
        en_text = clean_html(item.get("excerpt", ""))
        if en_text:
            all_rows.append(
                {
                    "Domain": "Security and Safety",
                    "Sub_Category": "Road Safety",
                    "English": en_text,
                    "Kiswahili": translate_text(en_text),
                    "Target_Language": "",
                    "Source": "NTSA (National Transport and Safety Authority)",
                    "Date": to_date(item.get("published_at", "")),
                    "Metadata": f"tone=advisory; platform=website; category={item.get('category', '')}",
                }
            )
except Exception as e:
    print(f" -> Error fetching NTSA data: {e}")

# =========================================================
# SECTION 2: NC4 (Cybersecurity)
# =========================================================
print("\n[2/4] Fetching data from NC4...")
url_nc4 = "https://studio.nc4.go.ke/api/news-items"
params_nc4 = {
    "filters[type][$eq]": "alert",
    "sort": "publishedAt:desc",
    "pagination[pageSize]": 100,
}

try:
    resp_nc4 = requests.get(url_nc4, params=params_nc4, headers=HEADERS)
    resp_nc4.raise_for_status()
    items_nc4 = resp_nc4.json().get("data", [])
    print(f" -> Got {len(items_nc4)} alert items from NC4")

    for item in items_nc4:
        excerpt = (
            get_field(item, "excerpt")
            or get_field(item, "description")
            or get_field(item, "summary")
        )
        en_text = clean_html(excerpt)
        if en_text:
            all_rows.append(
                {
                    "Domain": "Security and Safety",
                    "Sub_Category": "Cybersecurity",
                    "English": en_text,
                    "Kiswahili": translate_text(en_text),
                    "Target_Language": "",
                    "Source": "NC4 (National Computer and Cybercrime Coordination Committee)",
                    "Date": to_date(get_field(item, "publishedAt")),
                    "Metadata": "tone=advisory; platform=website; alert_type=cybersecurity",
                }
            )
except Exception as e:
    print(f" -> Error fetching NC4 data: {e}")

# =========================================================
# SECTION 3: ODPP (News & Updates)
# =========================================================
print("\n[3/4] Fetching data from ODPP News & Updates...")
odpp_api_url = "https://odpp.go.ke/wp-json/wp/v2/posts"
page = 1

while True:
    try:
        params = {"page": page, "per_page": 20, "_fields": "title,date,link"}
        resp = requests.get(
            odpp_api_url, headers=HEADERS, params=params, verify=False, timeout=10
        )

        if resp.status_code != 200:
            break

        posts = resp.json()
        if not posts:
            break

        print(f" -> Fetching ODPP News API page {page} ({len(posts)} posts)...")

        for post in posts:
            raw_title = BeautifulSoup(
                post.get("title", {}).get("rendered", ""), "html.parser"
            ).text
            en_text = clean_html(raw_title)

            if len(en_text) > 15:
                all_rows.append(
                    {
                        "Domain": "Security and Safety",
                        "Sub_Category": "Public Prosecution",
                        "English": en_text,
                        "Kiswahili": translate_text(en_text),
                        "Target_Language": "",
                        "Source": "ODPP (Office of the Director of Public Prosecutions)",
                        "Date": to_date(post.get("date", "")),
                        "Metadata": "tone=official; platform=website; category=news_update",
                    }
                )
        page += 1
    except Exception as e:
        print(f" -> Finished ODPP News fetching: {e}")
        break

# =========================================================
# SECTION 4: ODPP (Press Statements)
# =========================================================
print("\n[4/4] Fetching data from ODPP Press Statements...")
press_url = "https://odpp.go.ke/press-statements/"

try:
    resp_press = requests.get(press_url, headers=HEADERS, verify=False)
    soup_press = BeautifulSoup(resp_press.text, "html.parser")
    list_items = soup_press.find_all("li")

    ignored_phrases = [
        "vision, mission",
        "core values",
        "support@prosecutions.go.ke",
        "home",
        "about us",
        "media centre",
        "careers",
        "contact us",
    ]

    for li in list_items:
        raw_text = li.get_text(strip=True)
        lowered_text = raw_text.lower()

        if any(ignored in lowered_text for ignored in ignored_phrases):
            continue

        if len(raw_text) > 20:
            en_text = clean_html(raw_text)

            doc_date = ""
            link_tag = li.find("a")
            if link_tag and link_tag.get("href"):
                href = link_tag.get("href")
                full_doc_url = (
                    href
                    if href.startswith("http")
                    else f"https://odpp.go.ke/{href.lstrip('/')}"
                )

                if (
                    "odpp.go.ke" in full_doc_url
                    or full_doc_url.startswith("/")
                ):
                    print(f" -> Extracting PDF Date: {en_text[:35]}...")
                    doc_date = extract_date_from_pdf(full_doc_url)

            all_rows.append(
                {
                    "Domain": "Security and Safety",
                    "Sub_Category": "Public Prosecution",
                    "English": en_text,
                    "Kiswahili": translate_text(en_text),
                    "Target_Language": "",
                    "Source": "ODPP (Office of the Director of Public Prosecutions)",
                    "Date": doc_date,
                    "Metadata": "tone=official; platform=website; category=press_statement",
                }
            )
except Exception as e:
    print(f" -> Error fetching ODPP Press Statements: {e}")

# =========================================================
# SECTION 5: BUILD DATAFRAME & EXPORT TO EXCEL
# =========================================================
if all_rows:
    df = pd.DataFrame(all_rows)

    # Drop potential exact duplicates across English text entries
    df = df.drop_duplicates(subset=["English"]).reset_index(drop=True)

    # Inject contiguous sequence numbers (SEC_001, SEC_002, etc.) across all combined entries
    df.insert(0, "PSA_ID", [f"SEC_{i:03d}" for i in range(1, len(df) + 1)])

    # Enforce precise column ordering
    fields = [
        "PSA_ID",
        "Domain",
        "Sub_Category",
        "English",
        "Kiswahili",
        "Target_Language",
        "Source",
        "Date",
        "Metadata",
    ]
    df = df[fields]

    output_file = "master_psa_dataset.xlsx"

    # Save output spreadsheet with Excel lock fallback protection
    try:
        df.to_excel(output_file, index=False)
        print(
            f"\n SUCCESS! Combined {len(df)} total records into '{output_file}'"
        )
    except PermissionError:
        timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
        alt_output_file = f"master_psa_dataset_{timestamp}.xlsx"
        df.to_excel(alt_output_file, index=False)
        print(
            f"\n Notice: Primary Excel file was locked. Saved {len(df)} records to '{alt_output_file}'"
        )
else:
    print("\n No rows gathered from any source. Output file was not created.")
```
